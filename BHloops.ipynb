```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = "BHloops 2026 exercise.xlsx"

df = pd.read_excel(data)
```

```python
H1 = df["H (kA/m)"].dropna().to_numpy()
B1 = df["B (T)"].dropna().to_numpy()

H2 = df["H (kA/m).1"].dropna().to_numpy()
B2 = df["B (T).1"].dropna().to_numpy()

H31 = df["H (kA/m)"].dropna().to_numpy()
B31 = df["B (T)"].dropna().to_numpy()

H32 = df["H (kA/m).1"].dropna().to_numpy()
B32 = df["B (T).1"].dropna().to_numpy()
```
